# Summary Explanation of the Random Forrest Algorithm

Imagine a busy post office, where every piece of mail—each unique in its urgency and content—needs sorting into the correct box. This is akin to our task: sorting emails into Outlook categories using a classifier, a tool sharp enough to differentiate with precision.

In this bustling digital post office, our classifier isn't just any sorter—it's a Random Forest, a collective wisdom of decision trees. It learns from a myriad of examples, determining which emails belong in which Outlook category, akin to sorting letters into labeled boxes based on their content and importance.

But our emails are complex, coming in with various themes and topics, much like parcels wrapped in layers. To understand what's beneath, we use LDA embeddings, a technique that peels back these layers to reveal the core themes of each email, akin to scanning a package to learn its contents without opening it.

However, the flow of mail is uneven—important emails are rare, while spam floods in. To adjust, we rebalance our data, ensuring our Random Forest doesn't become overwhelmed by the common, learning to recognize and prioritize those rare, crucial messages.

Thus equipped, our classifier—our modern, digital sorter—navigates the continuous stream of emails. It ensures that every message, whether a quick note or an urgent alert, finds its way to the right category in Outlook, ready for action. This isn't just sorting mail; it's about making sure no important message is ever lost in the shuffle, a nod to efficiency in our ever-connected world.

# Train/Test Split and Rebalance Data

```python
X_train_bal, X_test, y_train_bal, y_test = train_test_split_utility(
    upsampling_method='adasyn', 
    use_lda_features=True
    )
```

# Optimize Params

```python
# Returns a dictionary of best parameters
best_params = optimize_rf(X_train_bal, y_train_bal) 
```

# Train Random Forest Classifier

```python
rf_clf = train_rf_classifier(
    X_train_bal, y_train_bal,
    n_estimators=int(best_params['params']['n_estimators']),
    max_depth=int(best_params['params']['max_depth']),
    min_samples_split=int(best_params['params']['min_samples_split'])
)
```

# Score the Classifier

```python
score_rf_classifier(X_train_bal, X_test, y_train_bal, y_test, rf_clf)
```
